**LISTAS** E *NUMPY* **ARRAYS**

As listas em Python são estruturas de dados básicas que podem conter elementos de diferentes tipos (inteiros, strings, outras listas, etc.). Já a Numpy (Numerical Python) é uma biblioteca do Python que fornece suporte a arrays multidimensionais, estruturas de dados mais avançadas e eficientes para cálculos numéricos.


**EXERCÍCIO 2**:
Transformar uma lista em array


```python
import numpy as carolzinha

fornecedores = ['maria da galvanoplastia', 'joão do escapamento', 'claudio pneus' ]

listaEmArray = carolzinha.array(fornecedores)

print("Lista De Fornecedores: ", fornecedores)
print("Array de Fornecedores: ", listaEmArray)


#Existe uma eficiência em um usar Array (Numpy) ao invés de Listas.
#Já foi comprovado por testes que a Array é significativamente mais eficiente que Listas
```
